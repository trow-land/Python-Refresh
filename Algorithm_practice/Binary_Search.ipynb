### Binary Search

Finds the position of a target in a sorted array by dividing the array in halv until the vlue is found



1) Divide search space into 2 halves by finding middle index

2) compare middle element of search space with the key

3) if key == middle element then the target is found and the process ends

4) if key != middle element, decide which half to use and carry on

5) process continues until key is found or search space is exhausted

```python
sorted_list = [1, 4, 7, 10, 14, 18, 22, 27, 31, 36, 40, 45, 50, 55, 60, 66, 72, 78, 84, 90, 96, 102, 108]
```

Iterative implementation

```python
def binary_search (arr, low, high, target):

    while low <= high:

        middle = low + (high - low) // 2

        if arr[middle] == target:
            return middle

        elif arr[middle] < target:
            low = middle + 1
        else:
            high = middle - 1

    return -1

```

```python
print(len(sorted_list) - 1)
print(binary_search(sorted_list, 0, len(sorted_list) - 1, 10))
print(binary_search(sorted_list, 0, len(sorted_list) - 1, 108))
print(binary_search(sorted_list, 0, len(sorted_list) - 1, 1))
print(binary_search(sorted_list, 0, len(sorted_list) - 1, 66))
```

Recursive Iteration

```python
def binary_search_recursive(arr, low, high, target):
    if low <= high:
        middle = low + (high - low) // 2

        if arr[middle] == target:
            return middle
        elif arr[middle] < target:
            return binary_search_recursive(arr, middle + 1, high, target)
        else:
            return binary_search_recursive(arr, low, middle - 1, target)

    return -1
```

```python
print(len(sorted_list) - 1)
print(binary_search_recursive(sorted_list, 0, len(sorted_list) - 1, 10))
print(binary_search_recursive(sorted_list, 0, len(sorted_list) - 1, 108))
print(binary_search_recursive(sorted_list, 0, len(sorted_list) - 1, 1))
print(binary_search_recursive(sorted_list, 0, len(sorted_list) - 1, 66))
```
